```python
import os
import sys
import yaml

# Set paths
PROJECT_PATH = '/content/drive/MyDrive/NLP-V2'
CONFIG_PATH = os.path.join(PROJECT_PATH, 'rex_ikea/configs/stage_b_sft.yaml')
COLAB_CONFIG_PATH = os.path.join(PROJECT_PATH, 'rex_ikea/configs/stage_b_sft_colab.yaml')

# Add project to python path
sys.path.append(PROJECT_PATH)

# Modify config for Colab
with open(CONFIG_PATH, 'r') as f:
    config = yaml.safe_load(f)

# Update data root to Colab path
config['data']['data_root'] = os.path.join(PROJECT_PATH, 'IKEA-Manuals-at-Work')

# Save new config
with open(COLAB_CONFIG_PATH, 'w') as f:
    yaml.dump(config, f)

print(f"Created Colab config at {COLAB_CONFIG_PATH}")
```

```python
# Run Training
!python "{PROJECT_PATH}/rex_ikea/scripts/train/train_stage_b_sft.py" \
    --config "{COLAB_CONFIG_PATH}"
```
